```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_diabetes

# Настройка стиля графиков
sns.set_theme(style="whitegrid")

# 1. Загрузка Diabetes
diabetes = load_diabetes(as_frame=True)
df_diabetes = diabetes.frame

print("=== Датасет: Diabetes ===")
print("Размерность:", df_diabetes.shape)
print("\nПервые 5 строк:")
display(df_diabetes.head())
print("\nПоследние 5 строк:")
display(df_diabetes.tail())
print("\nИнформация о типах данных:")
df_diabetes.info()
```

```python
print("=== Статистический анализ (Diabetes) ===")
# Описательные статистики
display(df_diabetes.describe())

# Проверка пропущенных значений
print("\nПропущенные значения:")
print(df_diabetes.isnull().sum())

# Проверка дубликатов
print(f"\nКоличество дубликатов: {df_diabetes.duplicated().sum()}")
# В этом датасете нет пропусков и дубликатов, но мы выполнили проверку как требует задание
```

```python
# Визуализация распределения признака 'bmi' (индекс массы тела) тремя способами
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Гистограмма
sns.histplot(df_diabetes['bmi'], kde=False, ax=axes[0], color='skyblue')
axes[0].set_title('Histplot (bmi)')

# 2. График плотности
sns.kdeplot(df_diabetes['bmi'], ax=axes[1], fill=True, color='green')
axes[1].set_title('Kdeplot (bmi)')

# 3. Скрипичная диаграмма
sns.violinplot(x=df_diabetes['bmi'], ax=axes[2], color='orange')
axes[2].set_title('Violinplot (bmi)')

plt.tight_layout()
plt.show()

# Boxplot для поиска аномалий по всем признакам
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_diabetes, orient="h")
plt.title("Boxplot всех признаков (поиск выбросов)")
plt.show()
```

```python
# Матрица корреляции
plt.figure(figsize=(10, 8))
corr_matrix = df_diabetes.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Матрица корреляции (Diabetes)")
plt.show()

# Парный график (Pairplot) для 3 самых важных признаков + целевая
# Берем s1, bmi, bp и целевую переменную (target)
sns.pairplot(df_diabetes[['bmi', 'bp', 's1', 'target']], diag_kind='kde', corner=True)
plt.show()
```

```python
# 2. Загрузка датасета Titanic
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
try:
    df_titanic = pd.read_csv(url)
    print("=== Датасет: Titanic ===")
    print("Размерность:", df_titanic.shape)
    print("\nПервые 5 строк:")
    display(df_titanic.head())
    print("\nИнформация о данных:")
    df_titanic.info()
except Exception as e:
    print("Ошибка загрузки по URL:", e)
    print("Попробуйте позже или используйте другой датасет")
```

```python
print("=== Анализ пропусков (Titanic) ===")
print(df_titanic.isnull().sum())

# Обработка пропусков:
# 1. Заполним пропуски в 'Age' медианой (числовые данные)
df_titanic['Age'] = df_titanic['Age'].fillna(df_titanic['Age'].median())

# 2. Заполним пропуски в 'Embarked' модой (категориальные данные)
df_titanic['Embarked'] = df_titanic['Embarked'].fillna(df_titanic['Embarked'].mode()[0])

# 3. Удалим колонку 'Cabin', так как там слишком много пропусков (более 70%)
df_titanic.drop('Cabin', axis=1, inplace=True)

print("\nПосле очистки, пропусков осталось:")
print(df_titanic.isnull().sum())

# Статистика для категориальных признаков
print("\nЧастота категорий в 'Sex':")
print(df_titanic['Sex'].value_counts())
```

```python
# Распределение возраста по классам и полу
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Гистограмма возраста
sns.histplot(data=df_titanic, x='Age', hue='Survived', multiple="stack", ax=axes[0])
axes[0].set_title('Распределение возраста (выжившие/погибшие)')

# Скрипичная диаграмма по классам
sns.violinplot(data=df_titanic, x='Pclass', y='Age', hue='Sex', split=True, ax=axes[1])
axes[1].set_title('Возраст по классам и полу')

plt.tight_layout()
plt.show()
```
